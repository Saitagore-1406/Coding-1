## ***Write a program which will find all such numbers which are divisible by 7 but are not a multiple of 3, between 2000 and 3200. Printed the output as a comma-separated sequence.***

```python
numbers = []

for i in range(2000, 3201):
    if i % 7 == 0 and i % 3 != 0:
        numbers.append(str(i))

print(",".join(numbers))
```

***Given a list of characters, convert it into string.
Example:
input    [“C”. “O”, “D”, “I”, “N”, “G”]
output    Coding***

```python
characters = ["C", "O", "D", "I", "N", "G"]

result = "".join(characters)

print(result)
```

***With a given integral number n, write a program to generate a dictionary that contains (i, i*i).
Example:
input:    5    
output:    {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}***

```python
n = int(input("Enter a number: "))

result = {}

for i in range(1, n + 1):
    result[i] = i * i

print(result)
```
